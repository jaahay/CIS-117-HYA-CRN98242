**Part 2**: Implementing the Handlers:
Implement your own version of the HTMLParser called MyHTMLParser that inherits from HTMLParser. Add a new method named frequency:

```python
from html.parser import HTMLParser
from collections import Counter
import string
import re
import sys

class MyHTMLParser(HTMLParser):
    """
    A subclass of HTMLParser that collects all text data,
    filters out tokens containing punctuation, and can report
    word frequencies and dump raw text to a file.
    """
    def __init__(self):
        super().__init__()
        self._words = []
        self.counter = Counter()
        self.translator = str.maketrans('', '', string.punctuation)

    def handle_data(self, data):
        """
        Called for each block of text between tags.
        Discards any token containing punctuation.
        """
        data_words = data.translate(self.translator).split()
        self._words.extend(data_words)
        self.counter.update(data_words)

    def frequency(self, n):
        """
        Print all words that occur at least n times, in
        alphabetical order with counts.
        """
        ret = {}
        for word, freq in self.counter.items():
          if(freq >= n):
            ret[word] = freq
        print(dict(sorted(ret.items())))

    def print_clean_data(self):
      """
      Prints the data without tags
      """
      print(self._words)

if __name__ == '__main__':
    # Fetch and parse the page again
    link = 'https://collegeofsanmateo.edu/wellnesscenter'
    from urllib.request import urlopen

    resp = urlopen(link)
    html = resp.read().decode().lower()
    main_regex = r"(?s)<main[^>]*>(.*?)</main>"
    main_search = re.search(main_regex, html)
    if not main_search:
        print("Didn't find body")
        sys.exit(0)

    main = main_search.group()
    # print(main)
    parser = MyHTMLParser()
    parser.feed(main)

    # Show frequency for words occurring >= 10 times
    parser.frequency(6)
    print(parser.counter.get("wellness"))
    parser.print_clean_data()
```
