# Task 1: Basic 'if-else' statement
## Write a function that checks if a given number is positive.
* If the number is positive, print "The number is positive."
* If the number is zero or negative, print "The number is not positive."

```python
def check_positive(number):
  if number > 0:
    print("The number is positive.")
  if number <= 0:
    print("The number is not positive.")
  pass

# Test the function
check_positive(5)  # Expected output: The number is positive.
check_positive(-3)  # Expected output: The number is not positive.
check_positive(0)   # Expected output: The number is not positive.
```


# Task 2: Using 'if' and 'else'
## Write a function that checks if a given number is even or odd.
* If the number is even, print "The number is even."
* If the number is odd, print "The number is odd."

```python
def check_even_odd(number):
    if number % 2 == 0:
      print("The number is even.")
    else:
      print("The number is odd.")
    pass

# Test the function
check_even_odd(10)  # Expected output: The number is even.
check_even_odd(7)   # Expected output: The number is odd.
```

# Task 3: Using 'if', 'elif', and 'else'
## Write a function that grades a student's score based on the following criteria:
* If the score is 90 or above, print "A"
* If the score is between 80 and 89, print "B"
* If the score is between 70 and 79, print "C"
* If the score is between 60 and 69, print "D"
* If the score is below 60, print "F"

```python
def grade_student(score):
  if score >= 90:
    print("A")
  elif score >= 80:
    print("B")
  elif score >= 70:
    print("C")
  elif score >= 60:
    print("D")
  elif score < 60:
    print("F")
    pass

# Test the function
grade_student(92)  # Expected output: A
grade_student(85)  # Expected output: B
grade_student(73)  # Expected output: C
grade_student(68)  # Expected output: D
grade_student(50)  # Expected output: F
```

# Task 4: Nested 'if' statements
## Write a function that checks if a number is divisible by both 2 and 3.
* If the number is divisible by both 2 and 3, print "Divisible by both."
* If the number is divisible only by 2, print "Divisible by 2."
* If the number is divisible only by 3, print "Divisible by 3."
* If the number is divisible by neither, print "Not divisible by 2 or 3."

```python
def check_divisibility(number):
  if number % 2 == 0:
    if number % 3 == 0:
      print("Divisible by both.")
    else:
      print("Divisible by 2.")
  elif number % 3 == 0:
    print("DIvisible by 3.")
  else:
    print("Not divisible by 2 or 3.")
  pass

# Test the function
check_divisibility(12)  # Expected output: Divisible by both.
check_divisibility(8)   # Expected output: Divisible by 2.
check_divisibility(9)   # Expected output: Divisible by 3.
check_divisibility(7)   # Expected output: Not divisible by 2 or 3.
```

# Task 5: Complex Conditionals
## Write a function that takes a year as input and checks if it is a leap year.
* A leap year is exactly divisible by 4 except for years that are exactly divisible by 100.
* However, years divisible by 400 are leap years.
* Print "Leap year" if the year is a leap year, otherwise print "Not a leap year."

```python
def check_leap_year(year):
  if year % 400 == 0 or ( year % 4 == 0 and year % 100 != 0):
    print("Leap year.")
  else:
    print("Not a leap year.")
  pass

# Test the function
check_leap_year(2000)  # Expected output: Leap year.
check_leap_year(1900)  # Expected output: Not a leap year.
check_leap_year(2012)  # Expected output: Leap year.
check_leap_year(2019)  # Expected output: Not a leap year.
```

## Write a function that takes three numbers as input and returns the largest of the three.
* You must use conditional statements to solve this problem.

```python
def find_largest(num1, num2, num3):
  max = num1
  if num2 > max:
    max = num2
  if num3 > max:
    max = num3
  print(max)
  pass

# Test the function
find_largest(5, 10, 3)  # Expected output: 10
find_largest(7, 2, 7)   # Expected output: 7
find_largest(-1, -5, -3) # Expected output: -1
```
