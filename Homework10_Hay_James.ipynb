# Recursive vs Iterative functions
Some problems are recursive in nature, for example:
- The Fibonacci series: Fib(1) = 1, Fib(2) = 1, Fib(n) = Fib(n-1) + Fib(n-2)
- Factorial: Factorial(n) = n * Factorial(n-1)

A recursive function is one that calls itself and it is a simple way of implementing a function for naturally recursive problems. However, recursive functions are computationally intense, requiring much more memory because of multiple function calls. For efficiency purposes, we oftentimes choose to implement the iterative equivalent instead.

```python
# Write a recursive implementation of a function that returns the factorial of
# number received as a parameter
def rec_fibo(n):
  if n == 1 or n == 2:
    return 1
  return rec_fibo(n-2) + rec_fibo(n-1)
```

```python
# Write an iterative implementation of a function that returns the factorial of
# number received as a parameter
def fib_iterative(nth):
    fib_arr = [None, 1, 1]
    while len(fib_arr) != nth + 1:
        fib_arr.append(fib_arr[-2] + fib_arr[-1])
    return fib_arr[-1]
```

# Measuring Time
## Fibonacci
Analize and run the following cel:

```python
import time
import matplotlib.pyplot as plt

# Timing the recursive Fibonacci function
def time_fibonacci_recursive(n):
    start_time = time.time()
    result = rec_fibo(n) # call
    end_time = time.time()
    return end_time - start_time

# Timing the iterative Fibonacci function
def time_fibonacci_iterative(n):
    start_time = time.time()
    result = fib_iterative(n) # call
    end_time = time.time()
    return end_time - start_time

# Arrays to store values
n_values = range(10, 31)  # Test from Fibonacci(10) to Fibonacci(30)
recursive_times = []
iterative_times = []

# Collect execution times
for n in n_values:
    one_recursive_time = time_fibonacci_recursive(n)
    one_iterative_time = time_fibonacci_iterative(n)

    recursive_times.append(one_recursive_time)
    iterative_times.append(one_iterative_time)

# Plotting the data
plt.figure(figsize=(10, 6))
plt.plot(n_values, recursive_times, label="Recursive Fibonacci", marker='o', color='r')
plt.plot(n_values, iterative_times, label="Iterative Fibonacci", marker='x', color='b')

plt.title("Execution Time of Fibonacci (Recursive vs Iterative)")
plt.xlabel("n")
plt.ylabel("Time (seconds)")
plt.legend()
plt.grid(True)
plt.show()
```
